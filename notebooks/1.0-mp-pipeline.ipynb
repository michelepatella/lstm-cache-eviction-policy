### 1. Data Preparation

Synthetic dataset is generated to simulate realistic access patterns, including short-term repetition, oscillating toggles, cyclic scanning, noisy distortion, Zipfian popularity, and long-term memory effects. ### 2. Data Preprocessing

Raw data is cleaned by removing missing values, and temporal features are extracted using cyclical encoding of timestamps. Locality-aware features, such as local frequency and recency, are computed over a rolling window.

```python
preprocess_data()
```

### 3. Validation

Time-series-aware, grid search cross-validation is applied to preserve temporal order, with early stopping to prevent overfitting. This step ensures the best hyperparameters selection before final model training.

```python
validate_model()
```

### 4. Training

The model is trained using an LSTM network for capturing temporal dependencies, with post-training optimizations such as pruning and quantization to improve efficiency.

```python
train_model()
```

### 5. Testing

Standalone evaluation is performed to assess predictive model performance using traditional metrics.

```python
test_model()
```
